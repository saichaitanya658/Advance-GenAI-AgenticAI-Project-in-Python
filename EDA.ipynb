# DATA CLEANING

```python
data['Name'] = data['Name'].str.replace(r"\W",'',regex=True)
```

```python
data
```

```python
data['Domain'] = data['Domain'].str.replace('\W','',regex = True)
```

```python
data
```

```python
data['Age'] = data['Age'].str.extract('(\d+)')
```

```python
data
```

```python
data['Location'] = data['Location'].str.replace('\W','',regex = True)
```

```python
data
```

```python
data['Salary']= data['Salary'].str.extract('(\d+)')
```

```python
data
```

```python
data['Exp'] = data['Exp'].str.extract('(\d+)')
```

```python
clean_data = data.copy()
```

```python
data.info()
```

# MISSING VALUE TREATMENT

```python
import numpy as np
```

```python
clean_data
```

```python
clean_data['Age']=clean_data['Age'].fillna(np.mean(pd.to_numeric(clean_data['Age'])))
```

```python
clean_data
```

```python
clean_data['Location'] = clean_data['Location'].fillna(clean_data['Location'].mode()[0])
```

```python
clean_data
```

```python
clean_data['Exp']=clean_data['Exp'].fillna(np.mean(pd.to_numeric(clean_data['Exp'])))
```

```python
clean_data
```

```python
clean_data.info()
```

```python
clean_data['Name'] = clean_data['Name'].astype('category')
```

```python
clean_data['Domain']=clean_data['Domain'].astype('category')
```

```python
clean_data['Location'] = clean_data['Location'].astype('category')
```

```python
clean_data.info()
```

```python
clean_data['Age']=clean_data['Age'].astype(int)
```

```python
clean_data['Salary']=clean_data['Salary'].astype(int)
```

```python
clean_data['Exp'] = clean_data['Exp'].astype(int)
```

```python
clean_data.info()
```

```python
clean_data.to_csv('cleandata.csv')
```
